#### 📑 Descripción de columnas del dataset CIC-IDS-2017
| **Columna**                                                     | **Descripción**                                                                         |
| --------------------------------------------------------------- | --------------------------------------------------------------------------------------- |
| **Destination Port**                                            | Puerto de destino del flujo (ej. 80 para HTTP, 443 para HTTPS).                         |
| **Flow Duration**                                               | Duración total del flujo en microsegundos.                                              |
| **Total Fwd Packets**                                           | Número total de paquetes enviados desde origen → destino (forward).                     |
| **Total Backward Packets**                                      | Número total de paquetes enviados desde destino → origen (backward).                    |
| **Total Length of Fwd Packets**                                 | Suma de longitudes (bytes) de todos los paquetes forward.                               |
| **Total Length of Bwd Packets**                                 | Suma de longitudes (bytes) de todos los paquetes backward.                              |
| **Fwd Packet Length (Max/Min/Mean/Std)**                        | Estadísticas de tamaño de los paquetes forward.                                         |
| **Bwd Packet Length (Max/Min/Mean/Std)**                        | Estadísticas de tamaño de los paquetes backward.                                        |
| **Flow Bytes/s**                                                | Tasa de transmisión en bytes por segundo.                                               |
| **Flow Packets/s**                                              | Tasa de transmisión en paquetes por segundo.                                            |
| **Flow IAT (Mean/Std/Max/Min)**                                 | Estadísticas de tiempo entre llegada de paquetes en todo el flujo (Inter Arrival Time). |
| **Fwd IAT (Total/Mean/Std/Max/Min)**                            | Estadísticas de tiempos entre paquetes forward.                                         |
| **Bwd IAT (Total/Mean/Std/Max/Min)**                            | Estadísticas de tiempos entre paquetes backward.                                        |
| **Fwd PSH Flags / Bwd PSH Flags**                               | Número de flags TCP PSH (push) forward / backward.                                      |
| **Fwd URG Flags / Bwd URG Flags**                               | Número de flags TCP URG (urgent) forward / backward.                                    |
| **Fwd Header Length / Bwd Header Length**                       | Tamaño total de cabeceras en forward / backward.                                        |
| **Fwd Packets/s / Bwd Packets/s**                               | Tasa de paquetes forward / backward por segundo.                                        |
| **Min / Max / Mean / Std / Variance Packet Length**             | Estadísticas globales de tamaños de paquetes del flujo.                                 |
| **FIN / SYN / RST / PSH / ACK / URG / CWE / ECE Flag Count**    | Contadores de flags TCP observados en el flujo.                                         |
| **Down/Up Ratio**                                               | Relación entre el número de paquetes downstream y upstream.                             |
| **Average Packet Size**                                         | Tamaño promedio de los paquetes del flujo.                                              |
| **Avg Fwd Segment Size / Avg Bwd Segment Size**                 | Tamaño promedio de segmentos en forward / backward.                                     |
| **Fwd Header Length.1**                                         | Duplicado de `Fwd Header Length` (en algunos CSV).                                      |
| **Fwd Avg Bytes/Bulk, Fwd Avg Packets/Bulk, Fwd Avg Bulk Rate** | Métricas de ráfagas (bulk) de datos forward.                                            |
| **Bwd Avg Bytes/Bulk, Bwd Avg Packets/Bulk, Bwd Avg Bulk Rate** | Métricas de ráfagas (bulk) de datos backward.                                           |
| **Subflow Fwd Packets / Bytes**                                 | Paquetes y bytes en subflujos forward.                                                  |
| **Subflow Bwd Packets / Bytes**                                 | Paquetes y bytes en subflujos backward.                                                 |
| **Init_Win_bytes_forward / backward**                           | Tamaño de ventana inicial TCP anunciada en forward / backward.                          |
| **act_data_pkt_fwd**                                            | Número de paquetes con datos enviados en forward.                                       |
| **min_seg_size_forward**                                        | Tamaño mínimo de segmento en forward.                                                   |
| **Active (Mean/Std/Max/Min)**                                   | Estadísticas de intervalos en que el flujo estuvo activo (transmitiendo).               |
| **Idle (Mean/Std/Max/Min)**                                     | Estadísticas de intervalos en que el flujo estuvo inactivo (sin transmisión).           |
| **Label**                                                       | Clase del tráfico: `BENIGN` o tipo de ataque (`DDoS`, `PortScan`, `Web Attack`, etc.).  |
